```python
def comparar(g,i=0):
    tam = len(g)
    igual = 0
    reposta = " "
    for i in range(len(g)):
        if(np.isclose(g[i], 0)):
            igual = igual + 1
            if(igual == len(g)):
                reposta = "igual"
                break
        else:
            reposta = "Não igual"
            i = i+1
    return reposta
```

```python
def conj_grad(x0,A,b):
    k = 0
    g = grad(x0)
    d = -g
    print(g)
    valor_conver = []
    while(k < 101):
        if(comparar(g) == "Não igual"):
            t_k = -1*np.dot(grad(x0).T,d)/(np.dot(np.dot(d.T,A),d))
            x_k =  x0 + np.dot(t_k,d)
            Beta = np.dot(np.dot(d.T,A),grad(x_k))/(np.dot(np.dot(d.T,A),d))
            d_k = -grad(x_k) + np.dot(Beta,d)
            x0 = x_k
            d = d_k
            g = grad(x_k)
            print(g)
        elif(k == 100):
            print("Não encontrou ponto de minimo")
            break
        else:
            valor_conver.append(x0)
            print(x0,"Ponto mínimo")
            break
        k += 1
    return valor_conver
```
